```python
import os
import copy
import random
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix
from sklearn.ensemble import RandomForestClassifier

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

TRAIN_CSV = "PMI_Class_Split/train_process_clean.csv"
VAL_CSV   = "PMI_Class_Split/val_process_clean.csv"
TEST_CSV  = "PMI_Class_Split/test_process_clean.csv"

TARGET_COL = "class_label_encoded"

SEED = 42
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
OUT_DIR = "model_comparison_clean"

# deep models
BATCH_SIZE = 32
EPOCHS = 100
LR = 3e-4
WEIGHT_DECAY = 1e-4
PATIENCE = 12

# mlp
MLP_HIDDEN_DIMS = [256, 128]
MLP_DROPOUT = 0.3

# transformer
D_TOKEN = 128
N_HEADS = 4
N_LAYERS = 2
TRANS_DROPOUT = 0.3

# weather columns
WEATHER_EXACT = {"Wind_Run", "wind_speed", "temp_high", "temp_low", "Heat_D-D", "Temp_Out"}
WEATHER_PREFIXES = ("wind_dir_", "season_")



def seed_everything(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

seed_everything(SEED)

train_df = pd.read_csv(TRAIN_CSV)
val_df   = pd.read_csv(VAL_CSV)
test_df  = pd.read_csv(TEST_CSV)

for name, df in [("train", train_df), ("val", val_df), ("test", test_df)]:
    if TARGET_COL not in df.columns:
        raise ValueError(f"{TARGET_COL} not found in {name} CSV")

common_cols = [c for c in train_df.columns if c in val_df.columns and c in test_df.columns]
train_df = train_df[common_cols].copy()
val_df   = val_df[common_cols].copy()
test_df  = test_df[common_cols].copy()

print("train shape:", train_df.shape)
print("val shape:", val_df.shape)
print("test shape:", test_df.shape)

print("train total nan:", train_df.isna().sum().sum())
print("val total nan:", val_df.isna().sum().sum())
print("test total nan:", test_df.isna().sum().sum())

if train_df.isna().sum().sum() > 0 or val_df.isna().sum().sum() > 0 or test_df.isna().sum().sum() > 0:
    raise ValueError("NaNs still exist in CSVs. Clean them before training.")



feature_cols = [c for c in train_df.columns if c != TARGET_COL]

img_cols = [c for c in feature_cols if c.startswith("img_f")]
weather_cols = [
    c for c in feature_cols
    if (c in WEATHER_EXACT) or c.startswith(WEATHER_PREFIXES)
]

if len(img_cols) == 0:
    raise ValueError("No image columns found.")

print("\nFeature counts")
print("image:", len(img_cols))
print("weather:", len(weather_cols))



def build_feature_matrix(df, cols):
    if len(cols) == 0:
        return np.zeros((len(df), 1), dtype=np.float32)
    return df[cols].to_numpy(dtype=np.float32)

def prepare_flat_features(train_df, val_df, test_df, cols):
    X_train = build_feature_matrix(train_df, cols)
    X_val   = build_feature_matrix(val_df, cols)
    X_test  = build_feature_matrix(test_df, cols)

    y_train = train_df[TARGET_COL].to_numpy(dtype=np.int64)
    y_val   = val_df[TARGET_COL].to_numpy(dtype=np.int64)
    y_test  = test_df[TARGET_COL].to_numpy(dtype=np.int64)

    scaler = StandardScaler()
    X_train = scaler.fit_transform(X_train).astype(np.float32)
    X_val   = scaler.transform(X_val).astype(np.float32)
    X_test  = scaler.transform(X_test).astype(np.float32)

    return X_train, X_val, X_test, y_train, y_val, y_test

def compute_metrics(y_true, y_pred):
    return {
        "acc": accuracy_score(y_true, y_pred),
        "macro_f1": f1_score(y_true, y_pred, average="macro"),
    }


def run_random_forest(X_train, y_train, X_val, y_val, X_test, y_test, exp_dir):
    clf = RandomForestClassifier(
        n_estimators=400,
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        class_weight="balanced",
        random_state=SEED,
        n_jobs=-1,
    )

    clf.fit(X_train, y_train)

    val_pred = clf.predict(X_val)
    test_pred = clf.predict(X_test)

    val_metrics = compute_metrics(y_val, val_pred)
    test_metrics = compute_metrics(y_test, test_pred)

    print("\n[Random Forest] VAL")
    print(classification_report(y_val, val_pred, digits=4))
    print(confusion_matrix(y_val, val_pred))

    print("\n[Random Forest] TEST")
    print(classification_report(y_test, test_pred, digits=4))
    print(confusion_matrix(y_test, test_pred))

    pd.DataFrame({"y_true": y_test, "y_pred": test_pred}).to_csv(
        os.path.join(exp_dir, "rf_test_predictions.csv"), index=False
    )

    return {
        "model": "random_forest",
        "best_val_f1": val_metrics["macro_f1"],
        "best_epoch": "N/A",
        "test_acc": test_metrics["acc"],
        "test_macro_f1": test_metrics["macro_f1"],
    }


class FlatDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

class MLPClassifier(nn.Module):
    def __init__(self, input_dim, num_classes, hidden_dims=(256, 128), dropout=0.3):
        super().__init__()
        layers = []
        prev = input_dim
        for h in hidden_dims:
            layers.extend([
                nn.Linear(prev, h),
                nn.ReLU(),
                nn.Dropout(dropout),
            ])
            prev = h
        layers.append(nn.Linear(prev, num_classes))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        return self.net(x)

def run_epoch_mlp(model, loader, criterion, optimizer=None):
    is_train = optimizer is not None
    model.train() if is_train else model.eval()

    total_loss = 0.0
    all_preds, all_targets = [], []

    for X, y in loader:
        X = X.to(DEVICE)
        y = y.to(DEVICE)

        with torch.set_grad_enabled(is_train):
            logits = model(X)
            loss = criterion(logits, y)

            if is_train:
                optimizer.zero_grad()
                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

        total_loss += loss.item() * y.size(0)
        preds = torch.argmax(logits, dim=1)
        all_preds.extend(preds.detach().cpu().numpy())
        all_targets.extend(y.detach().cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_targets, all_preds)
    macro_f1 = f1_score(all_targets, all_preds, average="macro")
    return avg_loss, acc, macro_f1, np.array(all_targets), np.array(all_preds)

def run_mlp(X_train, y_train, X_val, y_val, X_test, y_test, exp_dir):
    train_ds = FlatDataset(X_train, y_train)
    val_ds   = FlatDataset(X_val, y_val)
    test_ds  = FlatDataset(X_test, y_test)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    model = MLPClassifier(
        input_dim=X_train.shape[1],
        num_classes=len(np.unique(y_train)),
        hidden_dims=MLP_HIDDEN_DIMS,
        dropout=MLP_DROPOUT,
    ).to(DEVICE)

    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

    best_state = None
    best_val_f1 = -1.0
    best_epoch = -1
    patience_counter = 0

    for epoch in range(1, EPOCHS + 1):
        train_loss, train_acc, train_f1, _, _ = run_epoch_mlp(model, train_loader, criterion, optimizer)
        val_loss, val_acc, val_f1, _, _ = run_epoch_mlp(model, val_loader, criterion, optimizer=None)

        print(
            f"[MLP] Epoch {epoch:03d} | "
            f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} train_f1={train_f1:.4f} | "
            f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} val_f1={val_f1:.4f}"
        )

        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_epoch = epoch
            best_state = copy.deepcopy(model.state_dict())
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= PATIENCE:
            break

    model.load_state_dict(best_state)

    test_loss, test_acc, test_f1, y_true, y_pred = run_epoch_mlp(model, test_loader, criterion, optimizer=None)

    print("\n[MLP] TEST")
    print(classification_report(y_true, y_pred, digits=4))
    print(confusion_matrix(y_true, y_pred))

    pd.DataFrame({"y_true": y_true, "y_pred": y_pred}).to_csv(
        os.path.join(exp_dir, "mlp_test_predictions.csv"), index=False
    )

    return {
        "model": "mlp",
        "best_val_f1": best_val_f1,
        "best_epoch": best_epoch,
        "test_acc": test_acc,
        "test_macro_f1": test_f1,
    }

class FusionDataset(Dataset):
    def __init__(self, x_img, x_weather, y):
        self.x_img = torch.tensor(x_img, dtype=torch.float32)
        self.x_weather = torch.tensor(x_weather, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.long)

    def __len__(self):
        return len(self.y)

    def __getitem__(self, idx):
        return {
            "image": self.x_img[idx],
            "weather": self.x_weather[idx],
            "target": self.y[idx],
        }

class ModalityProjector(nn.Module):
    def __init__(self, in_dim, out_dim, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(in_dim, out_dim),
            nn.LayerNorm(out_dim),
            nn.GELU(),
            nn.Dropout(dropout),
        )

    def forward(self, x):
        return self.net(x)

class FusionTransformer(nn.Module):
    def __init__(self, img_dim, weather_dim, use_weather, num_classes,
                 d_token=128, n_heads=4, n_layers=2, dropout=0.3):
        super().__init__()
        self.use_weather = use_weather

        self.img_proj = ModalityProjector(img_dim, d_token, dropout)
        if self.use_weather:
            self.weather_proj = ModalityProjector(weather_dim, d_token, dropout)

        n_tokens = 2 + (1 if self.use_weather else 0)  # CLS + image (+ weather)

        self.cls_token = nn.Parameter(torch.zeros(1, 1, d_token))
        self.pos_embed = nn.Parameter(torch.zeros(1, n_tokens, d_token))

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_token,
            nhead=n_heads,
            dim_feedforward=d_token * 4,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=True,
        )
        self.encoder = nn.TransformerEncoder(encoder_layer, num_layers=n_layers)

        self.head = nn.Sequential(
            nn.LayerNorm(d_token),
            nn.Linear(d_token, d_token),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_token, num_classes),
        )

        nn.init.trunc_normal_(self.cls_token, std=0.02)
        nn.init.trunc_normal_(self.pos_embed, std=0.02)

    def forward(self, x_img, x_weather=None):
        token_list = [self.img_proj(x_img).unsqueeze(1)]
        if self.use_weather:
            token_list.append(self.weather_proj(x_weather).unsqueeze(1))

        tokens = torch.cat(token_list, dim=1)
        B = x_img.size(0)
        cls = self.cls_token.expand(B, -1, -1)
        tokens = torch.cat([cls, tokens], dim=1)
        tokens = tokens + self.pos_embed[:, :tokens.size(1), :]

        enc = self.encoder(tokens)
        cls_out = enc[:, 0]
        return self.head(cls_out)

def prepare_fusion_features(train_df, val_df, test_df, use_weather):
    x_train_img = build_feature_matrix(train_df, img_cols)
    x_val_img   = build_feature_matrix(val_df, img_cols)
    x_test_img  = build_feature_matrix(test_df, img_cols)

    img_scaler = StandardScaler()
    x_train_img = img_scaler.fit_transform(x_train_img).astype(np.float32)
    x_val_img   = img_scaler.transform(x_val_img).astype(np.float32)
    x_test_img  = img_scaler.transform(x_test_img).astype(np.float32)

    if use_weather:
        x_train_weather = build_feature_matrix(train_df, weather_cols)
        x_val_weather   = build_feature_matrix(val_df, weather_cols)
        x_test_weather  = build_feature_matrix(test_df, weather_cols)

        weather_scaler = StandardScaler()
        x_train_weather = weather_scaler.fit_transform(x_train_weather).astype(np.float32)
        x_val_weather   = weather_scaler.transform(x_val_weather).astype(np.float32)
        x_test_weather  = weather_scaler.transform(x_test_weather).astype(np.float32)
    else:
        x_train_weather = np.zeros((len(train_df), 1), dtype=np.float32)
        x_val_weather   = np.zeros((len(val_df), 1), dtype=np.float32)
        x_test_weather  = np.zeros((len(test_df), 1), dtype=np.float32)

    y_train = train_df[TARGET_COL].to_numpy(dtype=np.int64)
    y_val   = val_df[TARGET_COL].to_numpy(dtype=np.int64)
    y_test  = test_df[TARGET_COL].to_numpy(dtype=np.int64)

    return x_train_img, x_val_img, x_test_img, x_train_weather, x_val_weather, x_test_weather, y_train, y_val, y_test

def run_epoch_transformer(model, loader, criterion, optimizer=None):
    is_train = optimizer is not None
    model.train() if is_train else model.eval()

    total_loss = 0.0
    all_preds, all_targets = [], []

    for batch in loader:
        x_img = batch["image"].to(DEVICE)
        x_weather = batch["weather"].to(DEVICE)
        y = batch["target"].to(DEVICE)

        with torch.set_grad_enabled(is_train):
            logits = model(x_img, x_weather)
            loss = criterion(logits, y)

            if is_train:
                optimizer.zero_grad()
                loss.backward()
                nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
                optimizer.step()

        total_loss += loss.item() * y.size(0)
        preds = torch.argmax(logits, dim=1)
        all_preds.extend(preds.detach().cpu().numpy())
        all_targets.extend(y.detach().cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_targets, all_preds)
    macro_f1 = f1_score(all_targets, all_preds, average="macro")
    return avg_loss, acc, macro_f1, np.array(all_targets), np.array(all_preds)

def run_transformer(train_df, val_df, test_df, use_weather, exp_dir):
    x_train_img, x_val_img, x_test_img, x_train_weather, x_val_weather, x_test_weather, y_train, y_val, y_test = \
        prepare_fusion_features(train_df, val_df, test_df, use_weather)

    train_ds = FusionDataset(x_train_img, x_train_weather, y_train)
    val_ds   = FusionDataset(x_val_img, x_val_weather, y_val)
    test_ds  = FusionDataset(x_test_img, x_test_weather, y_test)

    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader   = DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader  = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    model = FusionTransformer(
        img_dim=x_train_img.shape[1],
        weather_dim=x_train_weather.shape[1],
        use_weather=use_weather,
        num_classes=len(np.unique(y_train)),
        d_token=D_TOKEN,
        n_heads=N_HEADS,
        n_layers=N_LAYERS,
        dropout=TRANS_DROPOUT,
    ).to(DEVICE)

    criterion = nn.CrossEntropyLoss()
    optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

    best_state = None
    best_val_f1 = -1.0
    best_epoch = -1
    patience_counter = 0

    for epoch in range(1, EPOCHS + 1):
        train_loss, train_acc, train_f1, _, _ = run_epoch_transformer(model, train_loader, criterion, optimizer)
        val_loss, val_acc, val_f1, _, _ = run_epoch_transformer(model, val_loader, criterion, optimizer=None)

        print(
            f"[Transformer] Epoch {epoch:03d} | "
            f"train_loss={train_loss:.4f} train_acc={train_acc:.4f} train_f1={train_f1:.4f} | "
            f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} val_f1={val_f1:.4f}"
        )

        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_epoch = epoch
            best_state = copy.deepcopy(model.state_dict())
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= PATIENCE:
            break

    model.load_state_dict(best_state)

    test_loss, test_acc, test_f1, y_true, y_pred = run_epoch_transformer(model, test_loader, criterion, optimizer=None)

    print("\n[Transformer] TEST")
    print(classification_report(y_true, y_pred, digits=4))
    print(confusion_matrix(y_true, y_pred))

    pd.DataFrame({"y_true": y_true, "y_pred": y_pred}).to_csv(
        os.path.join(exp_dir, "transformer_test_predictions.csv"), index=False
    )

    return {
        "model": "transformer_fusion",
        "best_val_f1": best_val_f1,
        "best_epoch": best_epoch,
        "test_acc": test_acc,
        "test_macro_f1": test_f1,
    }


experiments = {
    "image_only": img_cols,
    "image_weather": img_cols + weather_cols,
}

os.makedirs(OUT_DIR, exist_ok=True)
all_results = []

for exp_name, cols in experiments.items():
    print(f"EXPERIMENT: {exp_name}")

    exp_dir = os.path.join(OUT_DIR, exp_name)
    os.makedirs(exp_dir, exist_ok=True)

    pd.DataFrame({"feature_name": cols}).to_csv(
        os.path.join(exp_dir, "used_features.csv"), index=False
    )

    # flat features for RF and MLP
    X_train, X_val, X_test, y_train, y_val, y_test = prepare_flat_features(
        train_df, val_df, test_df, cols
    )

    rf_result = run_random_forest(X_train, y_train, X_val, y_val, X_test, y_test, exp_dir)
    rf_result["experiment"] = exp_name
    all_results.append(rf_result)

    mlp_result = run_mlp(X_train, y_train, X_val, y_val, X_test, y_test, exp_dir)
    mlp_result["experiment"] = exp_name
    all_results.append(mlp_result)

    transformer_result = run_transformer(
        train_df, val_df, test_df,
        use_weather=(exp_name == "image_weather"),
        exp_dir=exp_dir
    )
    transformer_result["experiment"] = exp_name
    all_results.append(transformer_result)


results_df = pd.DataFrame(all_results)

results_df["best_epoch"] = results_df["best_epoch"].fillna("N/A")

results_df = results_df[
    ["experiment", "model", "best_val_f1", "best_epoch", "test_acc", "test_macro_f1"]
].sort_values(by=["experiment", "test_macro_f1"], ascending=[True, False])

results_df.to_csv(os.path.join(OUT_DIR, "summary.csv"), index=False)


print("FINAL SUMMARY")
print(results_df.to_string(index=False))
print(f"\nSaved to: {OUT_DIR}")
```
